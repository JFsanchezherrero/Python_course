## Ejemplo de BLAST utilizando Biopython para la secuencia en el fichero test.fasta

A partir de un fichero test.fasta, realiza una busqueda en la base de datos blastn de NCBI Blast

```python
## importa modulos
from Bio.Blast import NCBIWWW 
from Bio import SeqIO

## lee archivo input como Bio.Seq
record = SeqIO.read("data/test.fasta", format="fasta")

## Ejecuta BLAST
resultH = NCBIWWW.qblast("blastn", "nt", record.format("fasta"))

# output en XML por defecto
with open("results/blast_results.xml","w") as out_handle:
	 # resultH.read() sólo se puede ejecutar 1 vez!!
    out_handle.write(resultH.read())
resultH.close()
```

El comando anterior ha gnerado un archivo, blast_results.xml con la información de la búsqueda. Vamos a leerlo e importarlo

```python
#Lee un archivo XML con el resultado de una secuencia
result_handle = open("results/blast_results.xml")
from Bio.Blast import NCBIXML
blast_record = NCBIXML.read(result_handle)
```

```python
print(len(blast_record.alignments))
```

Hay un total de 50 resultados para esta búsqueda, almacenados en la lista: blast_record.alignments

Para cada alineamiento de blast_record.alignments, obten diferente información como: evalue, length

```python
print("\n## Alingments:")
for aln in blast_record.alignments:
    print("************************")
    print("sequence:", aln.title)
    print("length:", aln.length)
    print("accession:", aln.accession)
    print("************************")
    for hsp in aln.hsps:
        print("e value:", hsp.expect)
        print("score:", hsp.score)
        print("identities:", hsp.identities)
        print("Query Start:", hsp.query_start)
        print("Query End:", hsp.query_end)
        print("Subject Start:", hsp.sbjct_start)
        print("Subject End:", hsp.sbjct_end)
        print("Alingment Length:", hsp.align_length)
        print("Query\t" + hsp.query[0:50] + "...")
        print("Match\t" + hsp.match[0:50] + "...")
        print("Sbjct\t" + hsp.sbjct[0:50] + "...")

    print("##---------------------##")
```
