# Calculate Position Weight Matrix (PWM)

```python
## calculate PWM / PSWM / PSSM
import numpy as np

## get total substitutions
observed_frequencies = alignment.substitutions

## Subset as an example for better visualization
observed_frequencies = observed_frequencies.select("DCEHKR")
print("Observed")
print(format(observed_frequencies, "%.4f"))

## normalize the matrix:
observed_frequencies /= np.sum(observed_frequencies)
print("Normalized")
print(format(observed_frequencies, "%.4f"))

## relative frequency of occurrence of each residue:
residue_frequencies = np.sum(observed_frequencies, 0)
expected_frequencies = np.dot(
    residue_frequencies[:, None], residue_frequencies[None, :]
)
print("Expected")
print(format(expected_frequencies, "%.4f"))

## logarithm of the observed over the expeceted freq
PWM = np.log2(observed_frequencies / expected_frequencies)
print("PWM")
print(PWM)
```
