## What is Biopython?
Biopython is a collection of freely available Python (https://www.python.org) modules for computational molecular biology. Since its inception in 2000, Biopython has been continuously developed and maintained by a large group of volunteers worldwide.

The Biopython web site (http://www.biopython.org) provides an online resource for modules, scripts, and web links for developers of Python-based software for bioinformatics use and research. Biopython includes parsers for various bioinformatics file formats; access to online services; standard sequence class, sequence alignment and motif analysis tools; algorithmns for clustering, structural biology and phylogenetic analysis among others.

## What can I find in the Biopython package?

The main Biopython releases have lots of functionality, including:

- The ability to parse bioinformatics files into Python utilizable data structures, including support for the following formats:

    - Blast output – both from standalone and WWW Blast

    - Sequence formats: Clustalw; FASTA; GenBank

    - PubMed and Medline

    - ExPASy files, like Enzyme and Prosite

    - SCOP, including ‘dom’ and ‘lin’ files

- Files in the supported formats can be iterated over record by record or indexed and accessed via a Dictionary interface.

- Code to deal with popular on-line bioinformatics destinations such as:

    - NCBI – Blast, Entrez and PubMed services

    - ExPASy – Swiss-Prot and Prosite entries, as well as Prosite searches

- Interfaces to common bioinformatics programs such as: BLAST, CLUSTALW and EMBOSS command line tools

- A standard sequence class that deals with sequences, ids on sequences, and sequence features. Alonging to easily:

    - Perform common operations on sequences, such as translation, transcription and weight calculations.
    - Deal with alignments, including a standard way to create and deal with substitution matrices.


- Extensive documentation and help with using the modules, including this file, on-line wiki documentation, the web site, and the mailing list.

## Example usage of the Seq Class

As an example, let's save into a file within the same directory where this interactive python notebook is, an example fasta sequence or execute the following code to save it using python:

```python
seq_header = ">gi|2765658|emb|Z78533.1|CIZ78533 C.irapeanum 5.8S rRNA gene and ITS1 and ITS2 DNA"
seq_dna = "CGTAACAAGGTTTCCGTAGGTGAACCTGCGGAAGGATCATTGATGAGACCGTGGAATAAACGATCGAGTGAATCCGGAGGACCGGTGTACTCAGCTCACCGGGGGCATTGCTCCCGTGGTGACCCTGATTTGTTGTTG"
```

```python
# save in a file named test.fasta the seq_header and seq_dna strings
# use python with, open(), write()
with open("results/test_biopython.fasta", "w") as text_file:
    text_file.write(seq_header + "\n")
    text_file.write(seq_dna)    
```

You can see in the Jupyter dashboard or in your folder generated a new file name test_biopython.fasta

Now, we will loaded into a python again and converte into a python object, in this case a Bio.Seq object.

See additional details in https://biopython.org/docs/dev/api/Bio.Seq.html

```python
# loads SeqIO module from the BioPython package
from Bio import SeqIO
from Bio import Seq

# creates a seq_record o for each fasta within the test_biopython.fasta file
for seq_record in SeqIO.parse("results/test_biopython.fasta", "fasta"):
    ## seq_record.id contains the ID of the sequence
    print("## seq_record")
    print(seq_record)
```

As there is only one sequence, all information is stored in seq_record variable, which type is it?

```python
print("## type(seq_record)")
print(type(seq_record))
```

It is an special type of variable from Biopython module Bio.SeqRecord

Lets look a little more closely… and use Python’s dir() function to find out more about the SeqRecord object and what it does:




```python
dir(seq_record)
```

```python
## seq_record.id contains the fasta ID
print(seq_record.id)
```

```python
# seq_record.seq contains the sequence
print(seq_record.seq)

# we can calculate the length or other statistics
print(len(seq_record.seq))
```

```python
## we can also access to sequence by applying string methods to seq_record
## example
print(len(seq_record))
```

```python
# we can slice the sequence and maintain it as an object 
print(seq_record[10:15])
```

```python
# or we can convert to string
print(str(seq_record.seq)[10:15])
```

### Obtain complement, reverse complement, transcribed or translated sequences


```python
## we can get complement and reverse complement sequence
print("Original sequence record")
print(seq_record.seq)
```

```python
## Complement
print(Seq.complement(seq_record.seq))
```

```python
## Reverse complement
print(Seq.reverse_complement(seq_record.seq))
```

```python
## transcribed DNA into RNA
print(Seq.transcribe(seq_record.seq))
```

```python
## Or use python string method alternative replace()
str(seq_record.seq).replace("T", "U")
```

```python
## Translating the sequence to protein
Seq.translate(seq_record.seq)
```

## Example of Genbank retrieval

GenBank is the NIH genetic sequence database, an annotated collection of all publicly available DNA sequences. https://www.ncbi.nlm.nih.gov/genbank/

We can easily acces the database using Biopython

```python
genbank_acc='NC_045512.2'
print(genbank_acc)
```

Of course you can access the web and find details about the NC_045512.2 entry.
https://www.ncbi.nlm.nih.gov/nuccore/NC_045512.2

But what if you want programatic access? Or you are using program that requires to automatically download the information? You can use python and Biopython!

```python
## load Biopython module Entrez and SeqIO 
## that allow us to parse sequence records and enter the database
from Bio import SeqIO
from Bio import Entrez

## provide a valid email
Entrez.email = 'JoseFrancisco.Sanchez@uab.cat'

## fecth the database (nucleotide) with the accession of interest (genbank_acc)
## return text information (retmode="text") in genbank format (rettype="gb")
handle = Entrez.efetch(db='nucleotide', id=genbank_acc, rettype='gb', retmode='text')
## read the connection
GBseq = SeqIO.read(handle,'genbank')
handle.close() # close the connection
```

We have opened a connection with genbank using valid user id, we have fecthed the database (nucleotide) with the accession of interest (genbank_acc)
and have returned information in text (retmode="text") genbank format (rettype="gb"). 

Information is stored in GBseq object

```python
GBseq
```

```python
type(GBseq)
```

```python
## Show items stored in GBseq
print('Description:', GBseq.description)
print('Division:', GBseq.annotations['data_file_division'])
print('Species:', GBseq.annotations['source'])
print('Molecule:', GBseq.annotations['molecule_type'])
print('Sequence:', GBseq.seq[1:100], "...") ## just show the first 100 nucleotides
print('Length:',len(GBseq))
```

## Example of Genbank search

We might also search the database looking for genes of interest with a given function or in particular organism

```python
# load Biopython module Entrez
from Bio import Entrez

# Always tell NCBI who you are
Entrez.email = 'JoseFrancisco.Sanchez@uab.cat'

#Connect GenBank:
handle = Entrez.esearch(db="nucleotide", term="Arabidopsis[ORGN] AND topoisomerase[TITL]")
## provide the term of interest using appropiate tags [ORGN], [TITL]

# Save results (XML format as default)
# and close connection
results = Entrez.read(handle)
handle.close()
```

```python
## information is stoired in object results
print(results)
```

What type of object it is? 

```python
type(results)
```

It is a special Biopython dictionary and we can access keys, values and items

## Example of alignment

Now, we will finally see and example of an alingment and how to deal with it using Biopython.

Found in campus virtual a file named as sequences.aln, stored in Clustal format, download and place it in the same folder as this python notebook



We can read it into python by using the AlignIO package (https://biopython.org/wiki/AlignIO)

```python
# load Biopython AlignIO package
from Bio import AlignIO

# read file named 'sequences.aln' in clustal format, available in same folder as this python notebook
align = AlignIO.read('data/sequences.aln', 'clustal')
print(align)
```

```python
print("Alignment length %i" % align.get_alignment_length())
```

### Alignment conversion

There are multiple alignment formats and converting from one to other might be tedious. Using biopython is straightforward.

We will use the format() function applyed to the aling object

```python
## Convert the alignment into fasta format
print(format(align, "fasta"))
```

```python
# save the alignment into FASTA format and save in a file
AlignIO.write(align, 'sequencesAln.fa', 'fasta')
```

```python
print(format(align, "phylip"))
```

```python
print(format(align, "stockholm"))
```
